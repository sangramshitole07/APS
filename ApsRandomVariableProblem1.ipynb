Problem statement
An organization sends promotional emails. Historical data indicate that a customer clicks the link with probability p=0.30.
Define:


X = 1, customer clicks
    0, customer does not click
thus, X ~ Bernoulis(0.30)


1.What does X=1 represent?

X = 1 represents that the customer clicks the promotional link.

2. What does X = 0 represent?
X = 0 represents that the customer does not click the promotional link.

3. What are the possible values of X?

The possible values are 0 and 1.

4. Is X a discrete or continuous random variable?

X is a discrete random variable because it takes only two countable values.

5. Which probability distribution is suitable for X?

The Bernoulli distribution with parameter p = 0.30.

Step 2: Theoretical PMF

For a Bernoulli random variable:

P(X = x) = p^x (1 − p)^(1 − x),  x ∈ {0,1}.

Therefore, P(X = 0) = 0.70 and P(X = 1) = 0.30.

```python
import numpy as np
import matplotlib.pyplot as plt
p=0.30
sample_size=20
sample=np.random.binomial(n=1,p=p,size=sample_size)
print(sample)
```

1. How many observations were generated? 20.

2. What does each zero represent? A customer did not click.

3. What does each one represent? A customer clicked.

4. Are the generated observations the same as the theoretical probabilities? No. A finite random sample will generally not have exactly 30% clicks and 70% non-clicks.

```python
#count the number of clicks
number_of_clicks=np.sum(sample==1)
number_of_non_clicks=np.sum(sample==0)
print("Number of clicks : ",number_of_clicks)
print("Number of customers who did not click : ",number_of_non_clicks)
```

 4

yes

```python
#Step 5: Empirical PMF

#Empirical P(X = 0) = 0.80

#Empirical P(X = 1) = 0.20
```

Step 6: Compare empirical and theoretical PMFs

1. Is the empirical probability of a click exactly 0.30?

No. It is 0.20 for this sample.

2. Which outcome has the larger probability?

X = 0 (non-click) has the larger empirical probability, 0.80, and theoretically 0.70 > 0.30.

3. Why are empirical and theoretical probabilities different?

The theoretical PMF describes the long-run model probabilities, while the empirical PMF is calculated from only 20 random observations. Sampling variation causes differences.

```python
#Step 7: Plot empirical and theoretical PMFs



import matplotlib.pyplot as plt
x = [0, 1]
theoretical = [0.70, 0.30]
emp_p0 = number_of_non_clicks / sample_size
emp_p1 = number_of_clicks / sample_size
empirical = [emp_p0, emp_p1]
width = 0.35
plt.bar([i-width/2 for i in x], theoretical, width, label='Theoretical')
plt.bar([i+width/2 for i in x], empirical, width, label='Empirical')
plt.xticks(x, ['X=0', 'X=1'])
plt.ylabel('Probability')
plt.title('Theoretical vs Empirical PMF')
plt.legend()
plt.show()
```

For a Bernoulli variable, the empirical expectation is the empirical probability of \(X=1\):

$$ E[X] = 0.20 $$

And empirical variance:

$$ Var(X)=p(1-p)=0.20(0.80)=\boxed{0.16} $$

So the corrected values are 0.20 for expectation and 0.16 for variance.

### Step 13: Sample Size Impact

To demonstrate the Law of Large Numbers, we will vary the sample size for the Bernoulli experiment and observe how the empirical probabilities, mean, and variance approach their theoretical values.

| Sample size | Empirical P(X=0) | Empirical P(X=1) | Empirical mean | Empirical variance |
|-------------|------------------|------------------|----------------|--------------------|
| 10          |                  |                  |                |                    |
| 100         |                  |                  |                |                    |
| 1,000       |                  |                  |                |                    |
| 10,000      |                  |                  |                |                    |

```python
import numpy as np
import pandas as pd

p = 0.30
sample_sizes = [10, 100, 1000, 10000]

theoretical_p0 = 1 - p
theoretical_p1 = p
theoretical_mean = p
theoretical_variance = p * (1 - p)

results = []

for size in sample_sizes:
    sample = np.random.binomial(n=1, p=p, size=size)

    empirical_p0 = np.sum(sample == 0) / size
    empirical_p1 = np.sum(sample == 1) / size
    empirical_mean = np.mean(sample)
    empirical_variance = np.var(sample)

    results.append({
        'Sample size': size,
        'Empirical P(X=0)': f'{empirical_p0:.4f}',
        'Empirical P(X=1)': f'{empirical_p1:.4f}',
        'Empirical mean': f'{empirical_mean:.4f}',
        'Empirical variance': f'{empirical_variance:.4f}'
    })

df_results = pd.DataFrame(results)
display(df_results)

print(f"\nTheoretical values:\nP(X=0) = {theoretical_p0:.2f}\nP(X=1) = {theoretical_p1:.2f}\nMean = {theoretical_mean:.2f}\nVariance = {theoretical_variance:.2f}")
```

### Observation

As the sample size increases from 10 to 10,000, the empirical probabilities, mean, and variance generally become closer to their theoretical values of 0.70, 0.30, 0.30, and 0.21 respectively. This demonstrates that larger samples provide a better approximation to the theoretical distribution, illustrating the Law of Large Numbers.

## Problem Set 2: ML Use Case: Number of Fraudulent Transactions Flagged

**Problem statement:** A fraud-detection model examines 50 transactions every hour. Assume that each transaction has a 4% probability of being flagged as potentially fraudulent.

Let:
X = "number of transactions flagged in one hour"

Initially assume that individual flagging decisions are independent.
$X \sim \text{Binomial}(n=50, p=0.04)$

The fraud investigation team can examine at most four flagged transactions per hour.

**Find:**
1. Simulate the operation of the fraud-detection system for 20,000 independent one-hour periods.
   During every simulated hour:
   * exactly 50 transactions are examined;
   * each transaction has a flagging probability of 0.04;
   * X records the total number of flagged transactions.

**Hint:** use `np.random.binomial` function

```python
import numpy as np

# Define the parameters for the binomial distribution
n_transactions = 50  # Number of trials (transactions per hour)
p_flagged = 0.04     # Probability of a transaction being flagged
n_simulations = 20000 # Number of independent one-hour periods to simulate

# Simulate the number of flagged transactions for each one-hour period
# np.random.binomial(n, p, size) generates 'size' number of samples
# where each sample is the number of successes (flagged transactions)
# in 'n' trials (transactions) with 'p' probability of success.
flagged_transactions = np.random.binomial(n=n_transactions, p=p_flagged, size=n_simulations)

print(f"Simulated {n_simulations} one-hour periods.")
print(f"First 10 simulated values for flagged transactions: {flagged_transactions[:10]}")
print(f"Minimum flagged transactions in a period: {np.min(flagged_transactions)}")
print(f"Maximum flagged transactions in a period: {np.max(flagged_transactions)}")
```
