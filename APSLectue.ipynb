```python
n=5

prob_no_match = 1

for i in range(1,n):
    prob_no_match *= (365 - i) / 365

prob_match = 1 - prob_no_match

print("Probability of same birthday:", prob_match)

```

```python
import matplotlib.pyplot as plt

n_values = []
probabilities = []

# Iterate n from 1 to 100
for n in range(1, 365):
    if n == 0:
        prob_match = 0.0
    else:
        prob_no_match = 1
        for i in range(n):
            prob_no_match *= (365 - i) / 365
        prob_match = 1 - prob_no_match

    n_values.append(n)
    probabilities.append(prob_match)

# Plotting the results
plt.figure(figsize=(10, 5))
plt.plot(n_values, probabilities, marker='o', linestyle='-')
plt.title('Probability of Shared Birthday vs. Number of People')
plt.xlabel('Number of People (n)')
plt.ylabel('Probability of Shared Birthday')

plt.xticks(range(0, 101, 10))
plt.yticks([i/10 for i in range(11)])

plt.show()
```
