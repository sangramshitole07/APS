#Problem1: A fair coin is tossed 100 times.Calculate:

Number of heads  

Number of tails

Experimental probability of Head  

Experimental probability of Tail

```python
import random

# Number of coin tosses
num_tosses = 100

# Simulate coin tosses
tosses = [random.choice(['Heads', 'Tails']) for i in range(num_tosses)]

# Count heads and tails
num_heads = tosses.count('Heads')
num_tails = tosses.count('Tails')

# Calculate experimental probabilities
prob_heads = num_heads / num_tosses
prob_tails = num_tails / num_tosses

# Display the results
print(f"Number of tosses: {num_tosses}")
print(f"Number of Heads: {num_heads}")
print(f"Number of Tails: {num_tails}")
print(f"Experimental probability of Heads: {prob_heads:.2f}")
print(f"Experimental probability of Tails: {prob_tails:.2f}")
```

#Problem 2: check Effect of the number of trials

Repeat the coin-toss experiment for: n=10, 100, 1,000, 10,000

```python
import random

def simulate_coin_toss(num_tosses):
    """
    Simulates a coin toss experiment and calculates experimental probabilities.

    Args:
        num_tosses (int): The number of times to toss the coin.

    Returns:
        tuple: A tuple containing (num_heads, num_tails, prob_heads, prob_tails).
    """
    tosses = [random.choice(['Heads', 'Tails']) for _ in range(num_tosses)]
    num_heads = tosses.count('Heads')
    num_tails = tosses.count('Tails')
    prob_heads = num_heads / num_tosses
    prob_tails = num_tails / num_tosses
    return num_heads, num_tails, prob_heads, prob_tails

# Define the number of trials to test
trial_counts = [10, 100, 1000, 10000]

print("Effect of the number of trials on experimental probability:\n")

for n in trial_counts:
    heads, tails, p_heads, p_tails = simulate_coin_toss(n)
    print(f"For n = {n:,} tosses:")
    print(f"  Number of Heads: {heads}")
    print(f"  Number of Tails: {tails}")
    print(f"  Experimental probability of Heads: {p_heads:.2f}")
    print(f"  Experimental probability of Tails: {p_tails:.2f}\n")
```

### Observation:
As the number of trials (coin tosses) increases, the experimental probabilities of getting heads and tails tend to get closer to the theoretical probability of 0.5 for a fair coin. This means when we have a large sample space we will get mathematically correct value of the event


#Problem 3
Rolling a die: A fair die is rolled 6,000 times.

Calculate the experimental probability of each outcome: S={1,2,3,4,5,6}

```python
import random
from collections import Counter

# Number of die rolls
num_rolls = 6000

# Simulate die rolls
rolls = [random.randint(1, 6) for _ in range(num_rolls)]

# Count the occurrences of each outcome
outcome_counts = Counter(rolls)

# Calculate experimental probability for each outcome
print(f"Experimental probabilities for {num_rolls} die rolls:\n")
for outcome in sorted(outcome_counts.keys()):
    probability = outcome_counts[outcome] / num_rolls
    print(f"  Outcome {outcome}: Count = {outcome_counts[outcome]}, Probability = {probability:.4f}")

# Check if all outcomes were observed
if len(outcome_counts) < 6:
    print("\nNote: Not all outcomes (1-6) were observed in this simulation.")
```

# Problem 4 :
Addition rule for disjoint events A die is rolled 10,000 times. Define: A=getting 1, B= getting 6, check the events are joint or disjoint Verify: P(A∪B)=P(A)+P(B)

```python
import random
from collections import Counter

# Number of die rolls
num_rolls = 10000

# Simulate die rolls
rolls = [random.randint(1, 6) for _ in range(num_rolls)]

# Count the occurrences of each outcome
outcome_counts = Counter(rolls)

# Define events A and B
# Event A: getting a 1
# Event B: getting a 6

# Check if events A and B are disjoint
# Two events are disjoint if they cannot occur at the same time.
# Getting a 1 and getting a 6 on a single roll are mutually exclusive,
# so they are disjoint events.
print("Events A (getting a 1) and B (getting a 6) are disjoint.\n")

# Calculate P(A)
count_A = outcome_counts[1]
p_A = count_A / num_rolls
print(f"Experimental P(A) (getting a 1): {count_A}/{num_rolls} = {p_A:.4f}")

# Calculate P(B)
count_B = outcome_counts[6]
p_B = count_B / num_rolls
print(f"Experimental P(B) (getting a 6): {count_B}/{num_rolls} = {p_B:.4f}")

# Calculate P(A∪B) - the probability of getting a 1 OR a 6
# For disjoint events, P(A∪B) = P(A) + P(B)
# We can also calculate it directly by counting occurrences of 1 or 6.
count_A_union_B = outcome_counts[1] + outcome_counts[6]
p_A_union_B_direct = count_A_union_B / num_rolls
print(f"Experimental P(A∪B) (getting a 1 or 6 directly): {count_A_union_B}/{num_rolls} = {p_A_union_B_direct:.4f}")

p_A_plus_p_B = p_A + p_B
print(f"P(A) + P(B): {p_A:.4f} + {p_B:.4f} = {p_A_plus_p_B:.4f}\n")

# Verify P(A∪B) = P(A) + P(B)
if abs(p_A_union_B_direct - p_A_plus_p_B) < 0.001: # Allowing for minor floating point differences
    print("Verification successful: P(A∪B) ≈ P(A) + P(B) for disjoint events.")
else:
    print("Verification failed: P(A∪B) != P(A) + P(B).")
```

#Problem 5
Addition rule for overlapping events

For one die roll, define: A=getting an even number A={2,4,6} and B=getting a number greater than 3 B={4,5,6}

Verify: P(A∪B)=P(A)+P(B)−P(A∩B)

```python
import random
from collections import Counter

# Number of die rolls
num_rolls_overlapping = 10000

# Simulate die rolls
rolls_overlapping = [random.randint(1, 6) for _ in range(num_rolls_overlapping)]

# Count the occurrences of each outcome
outcome_counts_overlapping = Counter(rolls_overlapping)

# Define events A and B
# Event A: getting an even number A={2,4,6}
# Event B: getting a number greater than 3 B={4,5,6}

# Calculate P(A)
count_A_overlapping = outcome_counts_overlapping[2] + outcome_counts_overlapping[4] + outcome_counts_overlapping[6]
p_A_overlapping = count_A_overlapping / num_rolls_overlapping
print(f"Experimental P(A) (getting an even number): {count_A_overlapping}/{num_rolls_overlapping} = {p_A_overlapping:.4f}")

# Calculate P(B)
count_B_overlapping = outcome_counts_overlapping[4] + outcome_counts_overlapping[5] + outcome_counts_overlapping[6]
p_B_overlapping = count_B_overlapping / num_rolls_overlapping
print(f"Experimental P(B) (getting a number > 3): {count_B_overlapping}/{num_rolls_overlapping} = {p_B_overlapping:.4f}")

# Calculate P(A∩B) - intersection: getting an even number AND a number > 3 (i.e., 4 or 6)
# A∩B = {4,6}
count_A_intersect_B = outcome_counts_overlapping[4] + outcome_counts_overlapping[6]
p_A_intersect_B = count_A_intersect_B / num_rolls_overlapping
print(f"Experimental P(A∩B) (getting 4 or 6): {count_A_intersect_B}/{num_rolls_overlapping} = {p_A_intersect_B:.4f}")

# Calculate P(A∪B) - union: getting an even number OR a number > 3 (i.e., 2, 4, 5, 6)
# A∪B = {2,4,5,6}
count_A_union_B_direct = outcome_counts_overlapping[2] + outcome_counts_overlapping[4] + outcome_counts_overlapping[5] + outcome_counts_overlapping[6]
p_A_union_B_direct = count_A_union_B_direct / num_rolls_overlapping
print(f"Experimental P(A∪B) (getting 2, 4, 5, or 6 directly): {count_A_union_B_direct}/{num_rolls_overlapping} = {p_A_union_B_direct:.4f}")

# Verify P(A∪B) = P(A) + P(B) - P(A∩B)
p_A_plus_p_B_minus_p_A_intersect_B = p_A_overlapping + p_B_overlapping - p_A_intersect_B
print(f"P(A) + P(B) - P(A∩B): {p_A_overlapping:.4f} + {p_B_overlapping:.4f} - {p_A_intersect_B:.4f} = {p_A_plus_p_B_minus_p_A_intersect_B:.4f}\n")

if abs(p_A_union_B_direct - p_A_plus_p_B_minus_p_A_intersect_B) < 0.001: # Allowing for minor floating point differences
    print("Verification successful: P(A∪B) ≈ P(A) + P(B) - P(A∩B) for overlapping events.")
else:
    print("Verification failed: P(A∪B) != P(A) + P(B) - P(A∩B).")
```

#Problem 6
Multiplication rule for independent events

A coin is tossed and a die is rolled 10,000 times.

Define:

obtaining Head  

obtaining an even number

Verify: P(A∩B)=P(A)P(B)

```python
import random
from collections import Counter

# Number of trials for coin toss and die roll
num_trials = 10000

# --- Simulate Coin Tosses ---
coin_tosses = [random.choice(['Heads', 'Tails']) for _ in range(num_trials)]
count_heads = coin_tosses.count('Heads')
p_A = count_heads / num_trials
print(f"Experimental P(A) (getting a Head): {count_heads}/{num_trials} = {p_A:.4f}")

# --- Simulate Die Rolls ---
die_rolls = [random.randint(1, 6) for _ in range(num_trials)]

# Event B: getting an even number (2, 4, 6)
count_even = sum(1 for roll in die_rolls if roll % 2 == 0)
p_B = count_even / num_trials
print(f"Experimental P(B) (getting an even number): {count_even}/{num_trials} = {p_B:.4f}")

# --- Simulate Combined Events for Intersection (A∩B) ---
# A∩B: getting a Head AND an even number
# We'll assume independence, so we can calculate this from the individual simulations.
# Or, for a direct simulation, we would need to pair each toss with a roll.
# For verification, we can count instances where both conditions are met.

# To simulate A∩B directly, we need to pair each coin toss with a die roll.
# Let's create paired outcomes.
combined_outcomes = []
for i in range(num_trials):
    coin_result = random.choice(['Heads', 'Tails'])
    die_result = random.randint(1, 6)
    combined_outcomes.append((coin_result, die_result))

count_A_intersect_B_direct = sum(1 for coin, die in combined_outcomes if coin == 'Heads' and die % 2 == 0)
p_A_intersect_B_direct = count_A_intersect_B_direct / num_trials
print(f"Experimental P(A∩B) (getting a Head AND an even number directly): {count_A_intersect_B_direct}/{num_trials} = {p_A_intersect_B_direct:.4f}")

# --- Verify P(A∩B) = P(A) * P(B) ---
product_p_A_p_B = p_A * p_B
print(f"P(A) * P(B): {p_A:.4f} * {p_B:.4f} = {product_p_A_p_B:.4f}\n")

if abs(p_A_intersect_B_direct - product_p_A_p_B) < 0.01: # Allowing for minor floating point differences, a larger threshold given two separate simulations
    print("Verification successful: P(A∩B) ≈ P(A) * P(B) for independent events.")
else:
    print("Verification failed: P(A∩B) != P(A) * P(B).")
```
